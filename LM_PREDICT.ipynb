```python
import boto3
import pandas as pd

# S3 bucket and file
bucket = "chunculead"
key = "customer_churn.csv"

# Create S3 client
s3 = boto3.client("s3")

# Get the file from S3
response = s3.get_object(
    Bucket=bucket,
    Key=key
)

# Read CSV into pandas
df = pd.read_csv(response["Body"])

# Display first 5 rows
df.head()
```
